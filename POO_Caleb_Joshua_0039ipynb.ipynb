### Conceptos de POO en python

### El prompt IA gemini:
Lenguaje Python, crear la clase "animal" con los atributos "nombre", "raza", crear el constructor animal, y las funciones ladrar(), correr(), crear los objetos pitbull, chihuahua, nivel principiante, no es interactivo, proporciona tu los datos

```python
# Definición de la clase
class Animal:
    # El constructor (__init__) inicializa los atributos del objeto
    def __init__(self, nombre, raza):
        self.nombre = nombre
        self.raza = raza

    # Función (método) para ladrar
    def ladrar(self):
        print(f"{self.nombre} dice: ¡Guau, guau! 😨")

    # Función (método) para correr
    def correr(self):
        print(f"{self.nombre} está corriendo velozmente. 🏃‍♂️")

    def Morder(self):
        print(f"{self.nombre} está murdiendo a la vecina. 😥")

# --- Creación de objetos (instancias) ---

# Creamos el objeto 'pitbull' con datos específicos
pitbull = Animal("Rocky", "Pitbull")

# Creamos el objeto 'chihuahua' con datos específicos
chihuahua = Animal("Chispa", "Chihuahua")
# Creamos el objeto "perro callejero" con datos
perro_callejero = Animal("Solovino", "Callejero")

# --- Uso de los objetos ---

print(f"Perro 1: {pitbull.nombre}, Raza: {pitbull.raza}")
pitbull.ladrar()
pitbull.correr()

print("-" * 20)

print(f"Perro 2: {chihuahua.nombre}, Raza: {chihuahua.raza}")
chihuahua.ladrar()
chihuahua.correr()

print("-" * 20)

print(f"Perro 3: {perro_callejero.nombre}, Raza: {perro_callejero.raza}")
perro_callejero.ladrar()
perro_callejero.correr()
perro_callejero.Morder()
print("Programa realizado por Caleb Dorado, NL = 14")
```
